# **Reading the bronze data** 

```python
customers = spark.read.parquet("Files/Bronze/customers.parquet")
# df now is a Spark DataFrame containing parquet data from "Files/Bronze/customers.parquet".
print(f"The total row count is: {customers.count()}")
print("======================")
display(customers.limit(5))
```

```python
orders = spark.read.parquet("Files/Bronze/orders.parquet")
# df now is a Spark DataFrame containing parquet data from "Files/Bronze/orders.parquet".

print(f"The total row count is: {orders.count()}")
print("======================")

display(orders.limit(5))

```

```python
payments = spark.read.parquet("Files/Bronze/payments.parquet")
# df now is a Spark DataFrame containing parquet data from "Files/Bronze/payments.parquet".

print(f"The total row count is: {payments.count()}")
print("======================")


display(payments.limit(5))

```

```python
support_tickets = spark.read.parquet("Files/Bronze/support_tickets.parquet")
# df now is a Spark DataFrame containing parquet data from "Files/Bronze/support_tickets.parquet".

print(f"The total row count is: {support_tickets.count()}")

display(support_tickets.limit(5))

```

```python
web_activities = spark.read.parquet("Files/Bronze/web_activities.parquet")
# df now is a Spark DataFrame containing parquet data from "Files/Bronze/web_activities.parquet".

print(f"The total row count is: {web_activities.count()}")


display(web_activities.limit(5))

```

# **Saving the data as Delta tables**

```python
customers.write.format("delta").mode("overwrite").saveAsTable("Customers")
orders.write.format("delta").mode("overwrite").saveAsTable("orders")
payments.write.format("delta").mode("overwrite").saveAsTable("payments")
support_tickets.write.format("delta").mode("overwrite").saveAsTable("support_tickets")
web_activities.write.format("delta").mode("overwrite").saveAsTable("web_activities")
```
